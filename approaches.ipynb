Polynomial Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# -----------------------------------
# Read CSV
# -----------------------------------
df = pd.read_csv(r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv")

# -----------------------------------
# Target column
# -----------------------------------
target = "TotalSlumsScore"

# -----------------------------------
# Feature columns
# -----------------------------------
features = [
    "temperature",
    "Forehead_Height_Y",
    "eyelt_mask_area_pixels",
    "eyelt_upper_eyelid_curvature",
    "eyelt_lower_eyelid_curvature",
    "eyert_eye_aspect_ratio",
    "eyert_eyelid_circularity",
    "eyert_eccentricity",
    "eyert_compactness",
    "eyert_extent",
    "eyert_upper_eyelid_curvature",
    "eyert_hu_moment_1",
    "eyert_hu_moment_2",
    "eyert_hu_moment_5"
]

# Remove duplicate feature names
features = list(dict.fromkeys(features))

# -----------------------------------
# Keep required columns
# -----------------------------------
cols = features + [target]

# Convert to numeric
for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove missing values
df_clean = df[cols].dropna()


# -----------------------------------
# Prepare Data
# -----------------------------------
X = df_clean[features]
y = df_clean[target]

# -----------------------------------
# Polynomial Features
# degree=2 means:
# x, x², x1*x2
# -----------------------------------
poly = PolynomialFeatures(degree=2, include_bias=False)

X_poly = poly.fit_transform(X)

# Get polynomial feature names
poly_feature_names = poly.get_feature_names_out(features)

# -----------------------------------
# Train Polynomial Regression
# -----------------------------------
model = LinearRegression()
model.fit(X_poly, y)

# -----------------------------------
# Polynomial Coefficients
# -----------------------------------
coeff_df = pd.DataFrame({
    "Feature": poly_feature_names,
    "Coefficient": model.coef_
})

# Show all columns and rows
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

print("\nPolynomial Feature Coefficients:")
print(coeff_df)

# -----------------------------------
# Intercept
# -----------------------------------
print("\nIntercept:")
print(model.intercept_)

# -----------------------------------
# R² Score
# -----------------------------------
r2 = model.score(X_poly, y)

print("\nPolynomial Regression R² Score:", r2)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.impute import SimpleImputer
import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ────────────────────────────────────────────────────────────────

# ← change to your CSV file path
CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv"
MRN_COLUMN = "MRN"                # ← change if your MRN column has a different name
TARGET = "TotalSlumsScore"

FEATURES = [
    "temperature",
    "Forehead_Height_Y",
    "eyelt_mask_area_pixels",
    "eyelt_upper_eyelid_curvature",
    "eyelt_lower_eyelid_curvature",
    "eyert_eye_aspect_ratio",
    "eyert_eyelid_circularity",
    "eyert_eccentricity",
    "eyert_compactness",
    "eyert_extent",
    "eyert_upper_eyelid_curvature",
    "eyert_hu_moment_1",
    "eyert_hu_moment_2",
    "eyert_hu_moment_5",
]

TARGET_MRNS = [
    "000-167", "000-168", "000-166", "000-169", "000-170",
    "000-171", "000-172", "000-173", "000-174", "000-175", "000-176",
]

# ── 2. LOAD DATA ─────────────────────────────────────────────────────────────

df = pd.read_csv(CSV_PATH)


# Verify required columns exist
missing_cols = [c for c in FEATURES + [TARGET, MRN_COLUMN] if c not in df.columns]
if missing_cols:
    raise ValueError(f"Missing columns in CSV: {missing_cols}")

# ── 3. SPLIT: rows to predict vs rows to train on ────────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)
df_predict = df[predict_mask].copy()
df_train   = df[~predict_mask].copy()


if len(df_train) == 0:
    raise ValueError("No training data available after filtering out target MRNs and missing targets.")

# ── 4. PREPARE FEATURES ──────────────────────────────────────────────────────

X_train = df_train[FEATURES]
y_train = df_train[TARGET]
X_pred  = df_predict[FEATURES]

# Impute missing values with column median
imputer = SimpleImputer(strategy="median")
X_train_imp = imputer.fit_transform(X_train)
X_pred_imp  = imputer.transform(X_pred)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_imp)
X_pred_scaled  = scaler.transform(X_pred_imp)

# ── 5. TRAIN & EVALUATE ──────────────────────────────────────────────────────

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_scaled, y_train, test_size=0.2, random_state=42
)

model = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    min_samples_split=2,
    random_state=42,
    n_jobs=-1,
)
model.fit(X_tr, y_tr)

y_val_pred = model.predict(X_val)
mae = mean_absolute_error(y_val, y_val_pred)
r2  = r2_score(y_val, y_val_pred)

print("\n── Validation Metrics ─────────────────────────────")
print(f"  MAE : {mae:.3f}")
print(f"  R²  : {r2:.3f}")
print("────────────────────────────────────────────────────\n")

# ── 6. PREDICT FOR TARGET MRNs ───────────────────────────────────────────────

predictions = model.predict(X_pred_scaled)

results = df_predict[[MRN_COLUMN]].copy()
results["predicted_totalslumscore"] = np.round(predictions, 2)

# If the ground truth exists for some of these MRNs, show it alongside
if df_predict[TARGET].notna().any():
    results["actual_totalslumscore"] = df_predict[TARGET].values

print("── Predictions ─────────────────────────────────────")
print(results.to_string(index=False))
print("────────────────────────────────────────────────────\n")

# ── 7. FEATURE IMPORTANCE ────────────────────────────────────────────────────

importance_df = pd.DataFrame({
    "feature"   : FEATURES,
    "importance": model.feature_importances_,
}).sort_values("importance", ascending=False)

print("── Feature Importance ──────────────────────────────")
print(importance_df.to_string(index=False))
print("────────────────────────────────────────────────────\n")

```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# NAIVE BAYES
from sklearn.naive_bayes import GaussianNB

# Metrics
from sklearn.metrics import (
    mean_absolute_error,
    r2_score,
    accuracy_score,
    classification_report
)

import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ─────────────────────────────────────────────
CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv" # ← change to your CSV file path

MRN_COLUMN = "MRN"
TARGET = "TotalSlumsScore"

FEATURES = [
    "temperature",
    "eyert_extent",
    "Forehead_Height_Y",
    "eyert_hu_moment_5"
]

TARGET_MRNS = [
    "000-167", "000-168", "000-166", "000-169", "000-170",
    "000-171", "000-172", "000-173", "000-174", "000-175",
    "000-176"
]

# ── 2. LOAD CSV ───────────────────────────────────────────

df = pd.read_csv(CSV_PATH)


# Check columns
missing_cols = [c for c in FEATURES + [TARGET, MRN_COLUMN] if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ── 3. SPLIT DATA ─────────────────────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)

df_predict = df[predict_mask].copy()
df_train = df[~predict_mask].copy()

# Remove missing target rows
df_train = df_train.dropna(subset=[TARGET])


# ── 4. FEATURES & TARGET ─────────────────────────────────

X_train = df_train[FEATURES]
X_pred  = df_predict[FEATURES]

y_train = df_train[TARGET]

# ---------------------------------------------------------
# IMPORTANT:
# Naive Bayes works for CLASSIFICATION
# Convert target into classes
# ---------------------------------------------------------

# Example:
# 0-10   -> Low
# 10-20  -> Medium
# 20+    -> High

y_train_class = pd.cut(
    y_train,
    bins=[-np.inf, 10, 20, np.inf],
    labels=["Low", "Medium", "High"]
)

# ── 5. HANDLE MISSING VALUES ─────────────────────────────

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train)
X_pred_imp  = imputer.transform(X_pred)

# ── 6. SCALE FEATURES ────────────────────────────────────

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imp)
X_pred_scaled  = scaler.transform(X_pred_imp)

# ── 7. TRAIN TEST SPLIT ──────────────────────────────────

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_scaled,
    y_train_class,
    test_size=0.2,
    random_state=42
)

# ── 8. TRAIN GAUSSIAN NAIVE BAYES ────────────────────────

model = GaussianNB()

model.fit(X_tr, y_tr)

# ── 9. VALIDATION ────────────────────────────────────────

y_val_pred = model.predict(X_val)

accuracy = accuracy_score(y_val, y_val_pred)

print("\n── Validation Metrics ─────────────────────")
print(f"Accuracy : {accuracy:.3f}")

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred))

# ── 10. PREDICT TARGET MRNs ──────────────────────────────

predictions = model.predict(X_pred_scaled)

results = df_predict[[MRN_COLUMN]].copy()

results["Predicted_Class"] = predictions

# Add actual target if available
if df_predict[TARGET].notna().any():
    results["Actual_TotalSlumsScore"] = df_predict[TARGET].values

print("\n── Predictions ────────────────────────────")
print(results.to_string(index=False))
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    mean_absolute_error,
    r2_score,
    accuracy_score
)

import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ────────────────────────────────────
CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv" # ← change to your CSV file path

MRN_COLUMN = "MRN"
TARGET = "TotalSlumsScore"

FEATURES = [
    "temperature",
    "eyert_extent",
    "Forehead_Height_Y",
    "eyert_hu_moment_5"
]

TARGET_MRNS = [
    "000-167", "000-168", "000-166",
    "000-169", "000-170", "000-171",
    "000-172", "000-173", "000-174",
    "000-175", "000-176"
]

# ── 2. LOAD DATA ──────────────────────────────────────────

df = pd.read_csv(CSV_PATH)


# Clean column names
df.columns = df.columns.str.strip()

# Check required columns
missing_cols = [c for c in FEATURES + [TARGET, MRN_COLUMN] if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ── 3. SPLIT TRAIN & PREDICT DATA ─────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)

df_predict = df[predict_mask].copy()
df_train = df[~predict_mask].copy()

# Remove missing target rows
df_train = df_train.dropna(subset=[TARGET])


if len(df_train) == 0:
    raise ValueError("No training data available")

# ── 4. FEATURES & TARGET ──────────────────────────────────

X_train = df_train[FEATURES]
y_train = df_train[TARGET]

X_pred = df_predict[FEATURES]

# ----------------------------------------------------------
# IMPORTANT:
# Convert numeric target into integer classes
# ----------------------------------------------------------

y_train_class = y_train.round().astype(int)

# ── 5. HANDLE MISSING VALUES ──────────────────────────────

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train)
X_pred_imp = imputer.transform(X_pred)

# ── 6. SCALE FEATURES ─────────────────────────────────────

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imp)
X_pred_scaled = scaler.transform(X_pred_imp)

# ── 7. TRAIN TEST SPLIT ───────────────────────────────────

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_scaled,
    y_train_class,
    test_size=0.2,
    random_state=42
)

# ── 8. TRAIN GAUSSIAN NAIVE BAYES ─────────────────────────

model = GaussianNB()

model.fit(X_tr, y_tr)

# ── 9. VALIDATION ─────────────────────────────────────────

y_val_pred = model.predict(X_val)

# Accuracy
accuracy = accuracy_score(y_val, y_val_pred)

# Numeric metrics
mae = mean_absolute_error(y_val, y_val_pred)
r2 = r2_score(y_val, y_val_pred)

print("\n── Validation Metrics ─────────────────────")
print(f"Accuracy : {accuracy:.3f}")
print(f"MAE      : {mae:.3f}")
print(f"R²       : {r2:.3f}")

# ── 10. PREDICT TARGET MRNs ───────────────────────────────

predictions = model.predict(X_pred_scaled)

results = df_predict[[MRN_COLUMN]].copy()

# Numeric prediction output
results["Predicted_TotalSlumsScore"] = predictions

# Add actual values if available
if df_predict[TARGET].notna().any():
    results["Actual_TotalSlumsScore"] = df_predict[TARGET].values

print("\n── Predictions ────────────────────────────")
print(results.to_string(index=False))

```

Piecewise Linear Regression

```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ─────────────────────────────────────────────

CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv" 

MRN_COLUMN = "MRN"
TARGET = "TotalSlumsScore"

# Add feature names
FEATURES = [
"temperature",
    "eyert_extent",
    "Forehead_Height_Y",
    "eyert_hu_moment_5"
]

# MRNs to predict
TARGET_MRNS = [
"000-167", "000-168", "000-166",
    "000-169", "000-170", "000-171",
    "000-172", "000-173", "000-174",
    "000-175", "000-176"
]

# Piecewise column
PIECEWISE_COLUMN = "Age"

# Age groups
BINS = [55, 65, 75, 85]

LABELS = [
    "55-65",
    "65-75",
    "75-85"
]

# ── 2. LOAD DATA ──────────────────────────────────────────

df = pd.read_csv(CSV_PATH)

# Clean column names
df.columns = df.columns.str.strip()

# Check required columns
required_cols = FEATURES + [
    TARGET,
    MRN_COLUMN,
    PIECEWISE_COLUMN
]

missing_cols = [c for c in required_cols if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ── 3. REMOVE MISSING TARGET ──────────────────────────────

df = df.dropna(subset=[TARGET])

# ── 4. CREATE AGE GROUPS ──────────────────────────────────

df["AgeGroup"] = pd.cut(
    df[PIECEWISE_COLUMN],
    bins=BINS,
    labels=LABELS,
    include_lowest=True
)

print("\nAge Group Counts:\n")
print(df["AgeGroup"].value_counts())

# ── 5. SPLIT TRAIN & PREDICT DATA ─────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)

df_predict = df[predict_mask].copy()
df_train = df[~predict_mask].copy()

# ── 6. TRAIN MODELS FOR EACH GROUP ────────────────────────

all_predictions = []

for group in LABELS:

    print(f"\n━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    print(f"Age Group : {group}")
    print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")

    # Group train data
    train_group = df_train[
        df_train["AgeGroup"] == group
    ]

    # Group predict data
    predict_group = df_predict[
        df_predict["AgeGroup"] == group
    ]

    print(f"Training rows   : {len(train_group)}")
    print(f"Prediction rows : {len(predict_group)}")

    # Skip empty groups
    if len(train_group) < 2:
        print("Skipped (not enough training rows)")
        continue

    # Features & target
    X_train = train_group[FEATURES]
    y_train = train_group[TARGET]

    # Fill missing values
    X_train = X_train.fillna(X_train.median())

    # Train model
    model = LinearRegression()

    model.fit(X_train, y_train)

    # Validation on same group
    y_pred_train = model.predict(X_train)

    mae = mean_absolute_error(y_train, y_pred_train)
    mse = mean_squared_error(y_train, y_pred_train)
    r2 = r2_score(y_train, y_pred_train)

    print(f"MAE : {mae:.3f}")
    print(f"MSE : {mse:.3f}")
    print(f"R²  : {r2:.3f}")

    # Predict MRNs
    if len(predict_group) > 0:

        X_pred = predict_group[FEATURES]

        X_pred = X_pred.fillna(X_train.median())

        predictions = model.predict(X_pred)

        result = pd.DataFrame({
            MRN_COLUMN:
                predict_group[MRN_COLUMN].values,

            "AgeGroup":
                predict_group["AgeGroup"].values,

            "Predicted_TotalSlumsScore":
                predictions
        })

        # Optional actual values
        if TARGET in predict_group.columns:
            result["Actual_TotalSlumsScore"] = (
                predict_group[TARGET].values
            )

        all_predictions.append(result)

# ── 7. COMBINE RESULTS ────────────────────────────────────

if len(all_predictions) > 0:

    final_results = pd.concat(
        all_predictions,
        ignore_index=True
    )

    print("\n── FINAL PREDICTIONS ─────────────────────")

    print(final_results.to_string(index=False))
```

Stability Selection (Subsampling Feature Selection)

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ─────────────────────────────────────────────

CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv" 
MRN_COLUMN = "MRN"
TARGET = "TotalSlumsScore"

# Add feature names here
FEATURES = [
"temperature",
    "eyert_extent",
    "Forehead_Height_Y",
    "eyert_hu_moment_5"
]

# MRNs to predict
TARGET_MRNS = [
"000-167", "000-168", "000-166",
"000-169", "000-170", "000-171",
"000-172", "000-173", "000-174",
"000-175", "000-176"
]

# Stability Selection Parameters
N_ITERATIONS = 100
SUBSAMPLE_SIZE = 0.7
ALPHA = 0.01

# Feature selection threshold
STABILITY_THRESHOLD = 0.6

# ── 2. LOAD DATA ──────────────────────────────────────────

df = pd.read_csv(CSV_PATH)


# Clean column names
df.columns = df.columns.str.strip()

# Check columns
missing_cols = [c for c in FEATURES + [TARGET, MRN_COLUMN]
                if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ── 3. SPLIT TRAIN & PREDICT DATA ─────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)

df_predict = df[predict_mask].copy()
df_train = df[~predict_mask].copy()

# Remove missing target rows
df_train = df_train.dropna(subset=[TARGET])


if len(df_train) == 0:
    raise ValueError("No training data available")

# ── 4. FEATURES & TARGET ──────────────────────────────────

X = df_train[FEATURES]
y = df_train[TARGET]

X_pred = df_predict[FEATURES]

# ── 5. HANDLE MISSING VALUES ──────────────────────────────

imputer = SimpleImputer(strategy="median")

X_imp = imputer.fit_transform(X)
X_pred_imp = imputer.transform(X_pred)

# ── 6. SCALE FEATURES ─────────────────────────────────────

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_imp)
X_pred_scaled = scaler.transform(X_pred_imp)

# ── 7. STABILITY SELECTION ────────────────────────────────

feature_counts = np.zeros(len(FEATURES))

print("\nRunning Stability Selection...")

for i in range(N_ITERATIONS):

    # Random subsample
    sample_idx = np.random.choice(
        len(X_scaled),
        size=int(len(X_scaled) * SUBSAMPLE_SIZE),
        replace=False
    )

    X_sub = X_scaled[sample_idx]
    y_sub = y.iloc[sample_idx]

    # Lasso model
    model = Lasso(alpha=ALPHA)

    model.fit(X_sub, y_sub)

    # Count selected features
    selected = np.abs(model.coef_) > 1e-5

    feature_counts += selected

# Stability score
stability_scores = feature_counts / N_ITERATIONS

# ── 8. CREATE FEATURE TABLE ───────────────────────────────

stability_df = pd.DataFrame({
    "Feature": FEATURES,
    "StabilityScore": stability_scores
})

stability_df["Selected"] = np.where(
    stability_df["StabilityScore"] >= STABILITY_THRESHOLD,
    "YES",
    "NO"
)

# Sort by stability
stability_df = stability_df.sort_values(
    by="StabilityScore",
    ascending=False
)

print("\n── Stability Selection Results ────────────")

print(stability_df.to_string(index=False))

# ── 9. SELECT IMPORTANT FEATURES ──────────────────────────

selected_features = stability_df[
    stability_df["Selected"] == "YES"
]["Feature"].tolist()

print("\n── Selected Features ──────────────────────")

print(selected_features)

if len(selected_features) == 0:
    raise ValueError("No stable features selected")

# ── 10. FINAL MODEL USING SELECTED FEATURES ───────────────

X_final = df_train[selected_features]

# Impute
X_final_imp = imputer.fit_transform(X_final)

# Scale
X_final_scaled = scaler.fit_transform(X_final_imp)

# Train-test split
X_tr, X_val, y_tr, y_val = train_test_split(
    X_final_scaled,
    y,
    test_size=0.2,
    random_state=42
)

# Final model
final_model = Lasso(alpha=ALPHA)

final_model.fit(X_tr, y_tr)

# Validation
y_val_pred = final_model.predict(X_val)

mae = mean_absolute_error(y_val, y_val_pred)
mse = mean_squared_error(y_val, y_val_pred)
r2 = r2_score(y_val, y_val_pred)

print("\n── Final Model Metrics ────────────────────")

print(f"MAE : {mae:.3f}")
print(f"MSE : {mse:.3f}")
print(f"R²  : {r2:.3f}")

# ── 11. PREDICT TARGET MRNs ───────────────────────────────

if len(df_predict) > 0:

    X_pred_final = df_predict[selected_features]

    # Impute + Scale
    X_pred_final_imp = imputer.transform(X_pred_final)
    X_pred_final_scaled = scaler.transform(X_pred_final_imp)

    predictions = final_model.predict(X_pred_final_scaled)

    results = df_predict[[MRN_COLUMN]].copy()

    results["Predicted_TotalSlumsScore"] = predictions

    # Add actual values if available
    if df_predict[TARGET].notna().any():
        results["Actual_TotalSlumsScore"] = df_predict[TARGET].values

    print("\n── Predictions ────────────────────────────")

    print(results.to_string(index=False))

 
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# ── 1. CONFIG ─────────────────────────────────────────────

CSV_PATH = r"C:\Users\Asus01\Desktop\yukthi\Approaches-Study\Study_data.csv"

MRN_COLUMN = "MRN"
TARGET = "TotalSlumsScore"

# Add feature column names here
FEATURES = [

"temperature",
    "eyert_extent",
    "Forehead_Height_Y",
    "eyert_hu_moment_5"
]

# MRNs to predict
TARGET_MRNS = [
"000-167", "000-168", "000-166",
"000-169", "000-170", "000-171",
"000-172", "000-173", "000-174",
"000-175", "000-176"
]

# Ridge Regularization Strength
ALPHA = 1.0

# Coefficient threshold for Zero Selection
ZERO_THRESHOLD = 0.01

# ── 2. LOAD DATA ──────────────────────────────────────────

df = pd.read_csv(CSV_PATH)


# Clean column names
df.columns = df.columns.str.strip()

# Check columns
missing_cols = [c for c in FEATURES + [TARGET, MRN_COLUMN]
                if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ── 3. SPLIT TRAIN & PREDICT DATA ─────────────────────────

predict_mask = df[MRN_COLUMN].isin(TARGET_MRNS)

df_predict = df[predict_mask].copy()
df_train = df[~predict_mask].copy()

# Remove missing target rows
df_train = df_train.dropna(subset=[TARGET])


if len(df_train) == 0:
    raise ValueError("No training data available")

# ── 4. FEATURES & TARGET ──────────────────────────────────

X_train = df_train[FEATURES]
y_train = df_train[TARGET]

X_pred = df_predict[FEATURES]

# ── 5. HANDLE MISSING VALUES ──────────────────────────────

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train)
X_pred_imp = imputer.transform(X_pred)

# ── 6. SCALE FEATURES ─────────────────────────────────────

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imp)
X_pred_scaled = scaler.transform(X_pred_imp)

# ── 7. TRAIN TEST SPLIT ───────────────────────────────────

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_scaled,
    y_train,
    test_size=0.2,
    random_state=42
)

# ── 8. TRAIN RIDGE REGRESSION ─────────────────────────────

model = Ridge(alpha=ALPHA)

model.fit(X_tr, y_tr)

# ── 9. VALIDATION ─────────────────────────────────────────

y_val_pred = model.predict(X_val)

mae = mean_absolute_error(y_val, y_val_pred)
mse = mean_squared_error(y_val, y_val_pred)
r2 = r2_score(y_val, y_val_pred)

print("\n── Validation Metrics ─────────────────────")

print(f"MAE : {mae:.3f}")
print(f"MSE : {mse:.3f}")
print(f"R²  : {r2:.3f}")

# ── 10. PREDICT TARGET MRNs ───────────────────────────────

if len(df_predict) > 0:

    predictions = model.predict(X_pred_scaled)

    results = df_predict[[MRN_COLUMN]].copy()

    results["Predicted_TotalSlumsScore"] = predictions

    # Add actual values if available
    if df_predict[TARGET].notna().any():
        results["Actual_TotalSlumsScore"] = df_predict[TARGET].values

    print("\n── Predictions ────────────────────────────")

    print(results.to_string(index=False))

else:
    print("\nNo MRNs found for prediction")

# ── 11. FEATURE COEFFICIENTS ──────────────────────────────

coeff_df = pd.DataFrame({
    "Feature": FEATURES,
    "Coefficient": model.coef_
})

# Zero Selection
coeff_df["Selected"] = np.where(
    np.abs(coeff_df["Coefficient"]) < ZERO_THRESHOLD,
    "ZERO / Removed",
    "Selected"
)

print("\n── Feature Coefficients ───────────────────")

print(
    coeff_df.sort_values(
        by="Coefficient",
        ascending=False
    ).to_string(index=False)
)

# ── 12. IMPORTANT FEATURES ONLY ───────────────────────────

important_features = coeff_df[
    np.abs(coeff_df["Coefficient"]) >= ZERO_THRESHOLD
]

print("\n── Important Features ─────────────────────")

print(
    important_features.sort_values(
        by="Coefficient",
        ascending=False
    ).to_string(index=False)
)

```
