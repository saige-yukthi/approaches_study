```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import statsmodels.api as sm

# -------------------------------
# Select required columns
# -------------------------------
cols = ["TotalSlumsScore", "OnThyroidMedicationSince"]

# Convert to numeric
df["TotalSlumsScore"] = pd.to_numeric(df["TotalSlumsScore"], errors="coerce")
df["OnThyroidMedicationSince"] = pd.to_numeric(df["OnThyroidMedicationSince"], errors="coerce")

# Remove missing values
df2 = df[cols].dropna()


# -------------------------------
# Pearson Correlation
# -------------------------------
pearson_corr, pearson_p = pearsonr(
    df2["TotalSlumsScore"],
    df2["OnThyroidMedicationSince"]
)

print("\n===== PEARSON CORRELATION =====")
print("Correlation Coefficient :", pearson_corr)
print("P-value :", pearson_p)

# -------------------------------
# Spearman Correlation
# -------------------------------
spearman_corr, spearman_p = spearmanr(
    df2["TotalSlumsScore"],
    df2["OnThyroidMedicationSince"]
)

print("\n===== SPEARMAN CORRELATION =====")
print("Correlation Coefficient :", spearman_corr)
print("P-value :", spearman_p)


```
